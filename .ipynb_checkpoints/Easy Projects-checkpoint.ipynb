```python
#Quiz Game

print("Welcome to the quiz!")

playing = input("Do you want to play? ").lower()

if playing != 'yes':
    quit()
    
print("Okay lets play!")

score = 0

answer1 = input("What is 2+2? ")

if answer1 == '4':
    print("Correct")
    score = score +1
else:
    print("Incorrect")
    
answer2 = input("What is 2+3? ")

if answer2 == '5':
    print("Correct")
    score = score +1
else:
    print("Incorrect")
    
answer3 = input("What is 2+4? ")

if answer3 == '6':
    print("Correct")
    score = score +1
else:
    print("Incorrect")
    
print("Your total score is:", (score/3)*100 )
```

```python
#Number Guessing Game

import random

top_of_range = input("Type a number: ")

if top_of_range.isdigit():
    top_of_range = int(top_of_range)
    
    if top_of_range <= 0:
        print('Please type a number larger than 0')
        quit()
else:
    print('Please type a number')
    quit()

r = random.randint(0, top_of_range)
guesses = 0
a = -1

while a!=r:
    guesses += 1
    a = input("please type your guess! ")
    if a.isdigit():
        a = int(a)
    else:
        print('Please type a number')
        continue
        
    if a > r:
        print("Your guess is high")
    elif a <r:
        print("Your guess is low")
    else:
        print("You are correct, you got it in", guesses ,"guesses" )
    
```

```python
#Rock Ppaer Scissor

user_wins = 0
comp_wins = 0

rounds = int(input("How many rounds do you want to play? "))


for i in range(rounds):
    user_input=input("Type Rock/Paper/Scissor/q ").lower()
    comp_input = random.choice(['rock','paper','scissor'])
    
    if user_input == 'q':
        quit()
    if user_input in ['rock','paper','scissor']:
        if user_input == comp_input:
            print("Its a tie")
        elif user_input == 'rock' and comp_input == 'scissor':
            print("User wins!")
            user_wins +=1
        elif user_input == 'scissor' and comp_input == 'paper':
            print("User wins!")
            user_wins +=1
        elif user_input == 'paper' and comp_input == 'rock':
            print("User wins!")
            user_wins +=1
        else:
            print("System wins system chose",comp_input, "user chose", user_input)
            comp_wins +=1


print("Of the",rounds,"rounds, the user won",user_wins,"and the system won",comp_wins
      
      
# wins = {('rock','scissors'), ('scissors','paper'), ('paper','rock')}
# if user_input == comp_input:
#     print("It's a tie.")
# elif (user_input, comp_input) in wins:
#     print("User wins!")
#     user_wins += 1
# else:
#     print("Computer wins!")
#     comp_wins += 1
        
# Rock beats Scissors
# Scissors beats Paper
# Paper beats Rock
```
